# Phase 2: Classical Machine Learning

In this phase, we load the processed and engineered dataset to train classical machine learning models for both **Regression** and **Classification** tasks. 
This forms the `M5` block of the Flow-Cast PRD.

The core logic has been abstracted into `src/ml_models.py` which handles:
1. **Time-based splitting:** Ensuring models are trained on past data and evaluated on future data.
2. **Baseline Training:** Linear Regression, Ridge, Lasso, ElasticNet, RandomForest, and GradientBoosting.
3. **Bayesian Optimization:** Optuna-based hyperparameter tuning for XGBoost and LightGBM.
4. **Evaluation:** Scoring on a strict hold-out test set (MAE for regression, Macro F1 for classification).

```python
import os
import sys
import pandas as pd
import warnings

warnings.filterwarnings('ignore')

# Add the src directory to the path so we can import our modules
sys.path.append(os.path.join(os.path.abspath(''), '..', 'src'))
from ml_models import run_regression, run_classification

# Load the dataset
data_path = os.path.join('..', 'data', 'Processed', 'featured_dataset.csv')
df = pd.read_csv(data_path)
df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df.dropna()

print(f"Dataset loaded with shape: {df.shape}")
```

## 1. Regression Tasks

Flow-Cast requires predicting continuous variables such as `traffic_volume` and `travel_time`.
We evaluate these models primarily on **Mean Absolute Error (MAE)**. A lower MAE means the predicted count/time is closer to the actual observed value.

```python
# 1.1 Traffic Volume Prediction
# Inference: Predicts the expected vehicle count for a given road segment in the next time window.
print("Starting Traffic Volume Regression Pipeline...")
run_regression(df, target_col='traffic_volume')
```

```python
# 1.2 Travel Time Prediction
# Inference: Estimates how long it will take to traverse a road segment.
print("Starting Travel Time Regression Pipeline...")
run_regression(df, target_col='travel_time')
```

## 2. Classification Tasks

Flow-Cast also requires predicting categorical risk levels such as `congestion_level` and `accident_count` (which represents accident risk). 
Because traffic anomalies and accidents are inherently imbalanced, we evaluate these models primarily on the **Macro F1-Score** rather than raw accuracy.

```python
# 2.1 Congestion Level Prediction
# Inference: Classifies the severity of congestion (e.g., Free-flow = 0, Moderate = 1, Heavy = 2, Severe = 3).
print("Starting Congestion Level Classification Pipeline...")
df['congestion_level'] = df['congestion_level'].astype(int)
run_classification(df, target_col='congestion_level')
```

```python
# 2.2 Accident Risk Prediction
# Inference: Classifies whether a segment is at high risk of an accident in the upcoming time window.
print("Starting Accident Risk Classification Pipeline...")
df['accident_count'] = df['accident_count'].astype(int)
run_classification(df, target_col='accident_count')
```
